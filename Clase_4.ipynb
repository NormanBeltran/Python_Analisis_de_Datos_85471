## Tabla de medidas estadísticas

| **Medida estadística**       | **Explicación sencilla**                                                                                     | **Ejemplo**                                                                                               |
| ---------------------------- | ------------------------------------------------------------------------------------------------------------ | --------------------------------------------------------------------------------------------------------- |
| **Media (promedio)**         | Es el valor que se obtiene al sumar todos los datos y dividirlos por la cantidad de datos.                   | Si los datos son 5, 7 y 8 → (5 + 7 + 8) / 3 = **6,67**                                                    |
| **Mediana**                  | Es el valor que queda justo en el medio cuando los datos están ordenados.                                    | Si los datos son 3, 7 y 9 → la mediana es **7**                                                           |
| **Moda**                     | Es el valor que más veces se repite.                                                                         | Si los datos son 2, 4, 4, 5 → la moda es **4**                                                            |
| **Rango**                    | Es la diferencia entre el valor mayor y el menor.                                                            | Si los datos son 2, 6, 9 → rango = 9 − 2 = **7**                                                          |
| **Varianza**                 | Mide cuánto se alejan los datos del promedio. Cuanto mayor sea, más dispersos están los datos.               | Si los datos son 2, 4, 6 → varianza = **2,67**                                                            |
| **Desvío estándar**          | Es la raíz cuadrada de la varianza. Indica, en promedio, cuánto se separan los datos del promedio.           | Si la varianza es 2,67 → desvío estándar = √2,67 ≈ **1,63**                                               |
| **Percentil / Cuartil**      | Indica la posición de un valor dentro del conjunto de datos. Por ejemplo, el percentil 50 es la mediana.     | En notas de un examen, estar en el percentil 90 significa que tu nota es mejor que el 90% de los alumnos. |
| **Coeficiente de variación** | Relación entre el desvío estándar y la media. Sirve para comparar la variabilidad entre conjuntos distintos. | Si media = 100 y desvío = 10 → 10/100 = **0,1 (10%)**                                                     |

## Vectorización

- Funcion poderosa que nos permite de manera sencilla aplicar operaciones sobre el array a todos los elementos

```python
ar1 = np.random.randint(1,100,size=36)
```

```python
ar1
```

```python
(ar1+10) * 1.2
```

```python
# Ejemplo de vectorización vs estructuras de python nativas

import numpy as np
import time

# Creamos dos vectores aleatorios
size = 5000
np.random.seed(0)
vec_1 = np.random.random(size)
vec_2 = np.random.random(size)

#____________________________________________________________________________________________________
# Implementación del producto vectorial en Python
tic = time.process_time()

python_product = np.zeros((size, size))

for i in range(size):
    for j in range(size):
        python_product[i][j] = vec_1[i] * vec_2[j]

phyton_time = time.process_time() - tic

#____________________________________________________________________________________________________
# Implementación del producto vectorial Numpy
tic = time.process_time()
numpy_product = np.outer(vec_1, vec_2)

numpy_time = time.process_time() - tic
#____________________________________________________________________________________________________

# Evaluación de resultados
if (numpy_product == python_product).all():
    print('Los resultados son iguales')
else:
    print('Los resultados son diferentes')
print('Tiempo de Python', phyton_time)
print('Tiempo de Numpy', numpy_time)
print('Mejora con Numpy', phyton_time / numpy_time)
```
