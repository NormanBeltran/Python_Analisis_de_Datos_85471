## Argumentos indeterminados (cantidad)

```python
def promedio(*args):
    print(args)
    total = sum(args)
    cantidad = len(args)
    return total / cantidad

print(promedio(1,2,3,4,5,6,7))
```

### Key Word Arguments (Diccionario y recibe claves y valores)

```python
def mostrar(**kwargs):
    print(kwargs)
    for k, v in kwargs.items():
        print(f"Clave {k} ==> Valor {v}")

mostrar(a=1, pepe="Juan", apellido="Perez", edad=34)
```

### Funciones de Orden Superior

- Todo en Python es objeto, incluso una funcion
- Aquella que recibe como parametro el nombre de otra funcion

```python
def cuadrado(x):
    return x**2

def cubo(x):
    return x**3

# Funcion de orden superior
def aplicar(lista, funcion):
    resultado = []
    for i in lista:
        resultado.append(funcion(i))
    return resultado

numeros=[1,2,3,4]

r = aplicar(numeros, cubo)
print(r)
```

### Map: Funcion de OS
- Primer argumento: Funcion
- Segundo argumento: Estructura iterable (lista, tupla, conjunto)

```python
def cuadrado(x):
    return x**2

def cubo(x):
    return x**3

numeros=[1,2,3,4]

cua = list(map(cuadrado, numeros))
cub = tuple(map(cubo, numeros))

print(cua, cub)
```

### Filter: Funcion de OS

```python
def bisiesto(anio):
    if anio%400==0:
        return True
    else:
        if anio%4==0 and anio%100!=0:
            return True
    return False

anios = [1800, 1972, 1999, 2020,2021, 1994,1976,1988]

res = list(filter(bisiesto, anios))
print(res)
```

### Ejemplo uso random como libreria

- choices con repeticion
- sample sin repeticion

```python
import random as rn

poblacion = ["Juan", "Ana", "Maria", "Pedro", "Marcela", "Mariana", "Susana", "Carlos"]
for i in range(10):
    print(rn.choices(poblacion, k=3))
```

```python
import random as rn

poblacion = ["Juan", "Ana", "Maria", "Pedro", "Marcela", "Mariana", "Susana", "Carlos"]
for i in range(10):
    print(rn.sample(poblacion, k=3))
```
