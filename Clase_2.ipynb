## Enumerate & Zip

```python
lista
```

```python
ord = 0

for i in lista:
    print(f"El indice de la posicion {ord} tiene {i}")
    ord += 1
```

```python
for idx, valor in enumerate(lista):
    print(f"El indice de la posicion {idx} tiene {valor}")
```

```python
nombres = ["ana", "pedro", "juan", "mateo", "maria", "juana"]
apellidos = ["perez", "gomez", "garcia", "rodriguez", "alvarez", "lopez"]

for n, a in zip(nombres, apellidos):
    print(f" Nombre: {n}, Apellido: {a}")
```

## Variante del While y For

- continue
- break

El else en un while o en un for se ejecuta si y solo si el bucle termina naturalmente (o por la condicion del while o por el fin del ciclo del for), si hay un break en algun ciclo no se ejecuta el else

```python
a = 1

while a < 10:
    #print(f"Antes del if {a}")
    
    if a == 3:
        a += 1
        continue

    if a == 4:
        break

    print(a)
    a += 1
else:
    print("Ingrese en el else")

print("Fin de programa")   
```

## Listas por comprension

- Aplica sobre cualquier coleccion mutable

```python
lista = []
for i in range(1, 101):
    lista.append(i)
```

```python
# lista = [ expresion    bucle    condiciones   ]

lista = [i   for i in range(1,101) ]
```

```python
lista
```

```python
# Diccionario de frecuencias

ausentes = ["ana", "ana", "maria", "juan", "juan", "juan", "juan", "pedro", "juan", "jose"]

diccionario = {}

for nombre in ausentes:
    if nombre in diccionario:
        diccionario[nombre] += 1
    else:
        diccionario[nombre] = 1

print(diccionario)
```

```python
dicc = { nombre : ausentes.count(nombre)       for nombre in set(ausentes)     } # Si es eficiente el algoritmo
```
